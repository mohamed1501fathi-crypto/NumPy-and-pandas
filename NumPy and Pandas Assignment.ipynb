```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# Convert Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)

# Find maximum, minimum and mean
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
print("First Three Days:")
print(temperatures_w1[:3])

print("Weekend Temperatures:")
print(temperatures_w1[-2:])

print("Middle Three Days:")
print(temperatures_w1[2:5])
```

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Two Weeks Temperatures:")
print(temperatures)
```

```python
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)

print("\nWeek 1 Temperatures:")
print(temperatures[0])

print("\nWeek 2 Temperatures:")
print(temperatures[1])

print("\nWeek 1 Weekend:")
print(temperatures[0, -2:])

print("\nWeek 2 Weekend:")
print(temperatures[1, -2:])
```

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

```python
print("First Rank Student Mark:")
print(marks[0])

print("\nTop Three Ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

print("\nThird Rank Student Mark:")
print(marks.iloc[2])

print("\nStudents with Marks Greater Than 90:")
print(marks[marks > 90])
```

```python
# Modify Rank1 mark to 100
marks.loc['Rank1'] = 100

print("After Modifying Rank1:")
print(marks)

# Remove the last rank
marks = marks.drop('Rank5')

print("\nAfter Removing Rank5:")
print(marks)

# Calculate CGPA
cgpa = marks / 10

print("\nCGPA:")
print(cgpa)
```

```python
import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],

    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print(transactions)
```

```python
# Display the DataFrame
print("Complete DataFrame:")
print(transactions)

# Display first 5 rows
print("\nHead:")
print(transactions.head())

# Display last 5 rows
print("\nTail:")
print(transactions.tail())

# Display shape
print("\nShape:")
print(transactions.shape)

# Display column names
print("\nColumn Names:")
print(transactions.columns)

# Display data types
print("\nData Types:")
print(transactions.dtypes)

# Display ProductCategory and Amount columns
print("\nSelected Columns:")
print(transactions[['ProductCategory', 'Amount']])

# Retrieve last 3 columns using iloc
print("\nLast Three Columns:")
print(transactions.iloc[:, -3:])

# Filter North region and Amount greater than 200
print("\nFiltered Transactions:")
print(transactions[(transactions['Region'] == 'North') &
                   (transactions['Amount'] > 200)])

# Count ProductCategory values
print("\nProduct Category Counts:")
print(transactions['ProductCategory'].value_counts())

# Find unique regions
print("\nUnique Regions:")
print(transactions['Region'].unique())

# Find mean amount for each region
print("\nMean Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# Modify Amount for TransactionID 102
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print("After Updating Amount:")
print(transactions)

# Add Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

print("\nAfter Adding Discount:")
print(transactions)

# Remove TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

print("\nAfter Removing TransactionID 109:")
print(transactions)

# Delete Discount column
transactions = transactions.drop(columns=['Discount'])

print("\nFinal DataFrame:")
print(transactions)
```
